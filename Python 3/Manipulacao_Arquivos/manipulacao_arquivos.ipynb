# *Manipulação de Arquivos*

### Leitura de arquivo simples

```python
#!/usr/bin/env python3
arquivo = open('pessoas.csv') #Abre o arquivo 'pessoas.csv' para leitura


dados = arquivo.read() #Lê o conteúdo do arquivo e armazena na variável 'dados'


arquivo.close() #Fecha o arquivo para liberar recursos do sistema


for registro in dados.splitlines():
    #print("Nome: {} | Idade: {}".format(*registro.split(','))) #O uso do * extrai os elementos da lista resultante de registro.split(',') e os passa como argumentos para format() para formatar a string de saída.   
    #print(registro)
    print(registro.split(','))
```

### Leitura de Arquivo como Stream e sua importância
    A leitura de arquivos em modo stream é essencial quando se trabalha com volumes grandes de dados ou sistemas que exigem eficiência de recursos. Nesse modelo, o arquivo não é carregado integralmente na memória; ele é processado gradualmente, geralmente linha a linha ou em blocos.

    Isso reduz drasticamente o consumo de RAM, tornando a aplicação mais escalável e estável, especialmente em ambientes com restrições de memória, como containers ou funções serverless. Além disso, permite processamento incremental, possibilitando validar, transformar ou interromper a leitura assim que uma condição for atendida, sem depender da carga completa do arquivo.

    Em termos de arquitetura, o uso de stream melhora a performance de I/O, evita gargalos e contribui para sistemas mais resilientes. Portanto, é uma prática recomendada em engenharia de dados, ETL e aplicações backend que manipulam arquivos extensos.

```python
#!/usr/bin/env python3
arquivo = open('pessoas.csv') #Abre o arquivo 'pessoas.csv' para leitura


for registro in arquivo.read().splitlines():
    print("Nome: {} | Idade: {}".format(*registro.strip().split(','))) #O uso do * extrai os elementos da lista resultante de registro.split(',') e os passa como argumentos para format() para formatar a string de saída.


arquivo.close() #Fecha o arquivo para liberar recursos do sistema
```

### Usando o bloco Try - Finally

```python
#!/usr/bin/env python3

try: #Usado para envolver o bloco de código que pode gerar uma exceção.
    arquivo = open('pessoas.csv')
    for registro in arquivo:
        print("Nome: {} | Idade: {}".format(*registro.strip().split(",")))
finally: #Este bloco será executado independentemente de uma exceção ocorrer ou não. Garante o fechamento do arquivo.
    arquivo.close()
    print("\033[31mArquivo fechado com sucesso!\033[0m") #A sequência de escape ANSI \033[31m é usada para definir a cor do texto para vermelho, e \033[0m é usada para resetar a formatação, garantindo que apenas a mensagem "Arquivo fechado com sucesso!" seja exibida em vermelho.
```

### Usando o bloco whith

```python
#!/usr/bin/env python3

# Usando a declaração with para garantir o fechamento automático do arquivo ou qualquer recurso, mesmo que ocorra uma exceção durante a execução do bloco de código.
with open('pessoas.csv') as arquivo:
    for registro in arquivo:
        print("Nome: {} | Idade: {}".format(*registro.strip().split(",")))


if arquivo.closed: #Verifica se o arquivo foi fechado
    print("\033[31mArquivo fechado com sucesso!\033[0m") #A sequência de escape ANSI \033[31m é usada para definir a cor do texto para vermelho, e \033[0m é usada para resetar a formatação, garantindo que apenas a mensagem "Arquivo fechado com sucesso!" seja exibida em vermelho.
```

### Escrevendo em um arquivo

```python
#!/usr/bin/env python3

# Usando a declaração with para garantir o fechamento automático do arquivo ou qualquer recurso, mesmo que ocorra uma exceção durante a execução do bloco de código.
with open('pessoas.csv') as arquivo:
    with open('pessoas.txt', 'w') as saida: #Abre o arquivo 'pessoas.txt' para escrita. Se o arquivo não existir, ele será criado. Se já existir, seu conteúdo será sobrescrito.
        for registro in arquivo:
            print("Nome: {} | Idade: {}".format(*registro.strip().split(",")), file=saida)


if arquivo.closed: #Verifica se o arquivo foi fechado
    print("\033[32mArquivo arquivo.csv fechado com sucesso!\033[0m") #A sequência de escape ANSI \033[31m é usada para definir a cor do texto para vermelho, e \033[0m é usada para resetar a formatação, garantindo que apenas a mensagem "Arquivo fechado com sucesso!" seja exibida em vermelho.


if saida.closed: #Verifica se o arquivo foi fechado
    print("\033[32mArquivo saida.txt fechado com sucesso!\033[0m") #A sequência de escape ANSI \033[31m é usada para definir a cor do texto para vermelho, e \033[0m é usada para resetar a formatação, garantindo que apenas a mensagem "Arquivo fechado com sucesso!" seja exibida em vermelho.
```
